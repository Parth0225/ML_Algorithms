## Ordinal Logistic Regression

### ex. Low , Medium , High

#### Ordered classes (Poor < Average < Good < Excellent)

```python
import numpy as np
import pandas as pd
from statsmodels.miscmodels.ordinal_model import OrderedModel

# 1. Generate a synthetic dataset
np.random.seed(42)
n = 500

x1 = np.random.randn(n)
x2 = np.random.randn(n)

# Create a continuous latent variable with some noise from a logistic distribution
latent = 0.8 * x1 - 1.5 * x2 + np.random.logistic(size=n)

# Discretize the latent variable into 3 ordered categories
y = pd.cut(latent, bins=[-np.inf, -0.5, 0.5, np.inf], labels=['Low', 'Medium', 'High'])
y = pd.Categorical(y, ordered=True)

df = pd.DataFrame({'y': y, 'x1': x1, 'x2': x2})

# 2. Fit the Ordinal Logistic Regression (Proportional Odds Model)
# Note: OrderedModel requires that there is no intercept (hence '0 +') 
# because category thresholds handle the intercepts.
mod = OrderedModel.from_formula("y ~ 0 + x1 + x2", df, distr="logit")
res = mod.fit(method="bfgs", disp=False)

# 3. Print summary of results
print(res.summary())

# 4. Predict probabilities for each category on new/existing data
predicted_probs = res.predict(df[['x1', 'x2']])
print("\nFirst 5 rows of predicted probabilities:")
print(predicted_probs.head())
```
