## Linear Regression

#### y=b0​+b1​x1​+b2​x2​

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# 1. Sample Data
# X: Study hours per week (Features)
X = np.array([[1.5], [2.0], [3.0], [4.0], [4.5], [5.5], [6.0], [7.0], [8.0], [9.0]])

# y: Exam marks obtained (Target)
y = np.array([45, 50, 60, 65, 70, 75, 80, 85, 92, 95])

# 2. Train the Model
model = LinearRegression()
model.fit(X, y)

# Print out the learned parameters
print(f"Slope (Marks gained per study hour): {model.coef_[0]:.2f}")
print(f"Intercept (Base marks with 0 study hours): {model.intercept_:.2f}")

# 3. Generate predictions for the line plot
# Create a smooth range of study hours from min to max to draw a clean line
X_range = np.linspace(X.min(), X.max(), 100).reshape(-1, 1)
y_pred = model.predict(X_range)

# 4. Visualization with Matplotlib
plt.figure(figsize=(9, 5))

# Plot actual student data points as a scatter plot
plt.scatter(X, y, color='blue', label='Actual Student Data', s=70, zorder=3)

# Plot the regression line learned by the model
plt.plot(X_range, y_pred, color='red', linewidth=2, label='Linear Regression Fit', zorder=2)

# Graph styling
plt.title('Relationship Between Study Hours and School Marks', fontsize=14, fontweight='bold', pad=12)
plt.xlabel('Study Hours per Week', fontsize=12)
plt.ylabel('Exam Marks (out of 100)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6, zorder=1)
plt.legend(fontsize=11)

# Show the plot
plt.tight_layout()
plt.show()
```
