## Ridge Regression

### Ridge regression tries to shrink the coefficients and also make sure not to remove / making them zero

#### Cost=RSS+λ(b12​+b22​+b32​+...)

```python
import numpy as np
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.preprocessing import StandardScaler

# 1. Create synthetic data with multicollinearity
np.random.seed(42)
n_samples = 100

study_hours = np.random.uniform(1, 10, n_samples)
# Video watch hours track very closely with study hours (heavy correlation)
video_hours = study_hours * 0.95 + np.random.normal(0, 0.1, n_samples) 
attendance = np.random.uniform(50, 100, n_samples)

# Target: Exam marks
marks = 20 + (4.0 * study_hours) + (3.0 * video_hours) + (0.3 * attendance) + np.random.normal(0, 2, n_samples)

# Combine features
X = np.column_stack([study_hours, video_hours, attendance])
feature_names = ['Study Hours', 'Video Hours (Collinear)', 'Attendance %']

# Scale features (Critical for Ridge, so penalties apply equally)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 2. Train Standard Linear Regression vs. Ridge Regression
lr_model = LinearRegression()
lr_model.fit(X_scaled, marks)

ridge_model = Ridge(alpha=10.0) # alpha controls regularization strength
ridge_model.fit(X_scaled, marks)

# 3. Compare the coefficients
print(f"{'Feature':<26} | {'Standard LR Coef':<18} | {'Ridge Coef (alpha=10)':<20}")
print("-" * 68)
for name, lr_c, ridge_c in zip(feature_names, lr_model.coef_, ridge_model.coef_):
    print(f"{name:<26} | {lr_c:<18.4f} | {ridge_c:<20.4f}")
```
