## KNN

### KNN stands for k nearest neighbors

### K - Nearest Neighbor is supervised learning algorithm that classifies a new data point based on its nearest neighbor. Classification is based on the majority voting

### we use euclidean distance to calculate the distance 

#### d = underroot(x2-x2)+(y2-y1)

```python
from sklearn.neighbors import KNeighborsClassifier

X = [
    [2,50],
    [3,60],
    [8,90],
    [9,95]
]

y = ['Fail','Fail','Pass','Pass']

model = KNeighborsClassifier(n_neighbors=3)

model.fit(X,y)

print(model.predict([[7,85]]))
```
