## Naive Bayes

### It is based on bayes theorem and work surprisingly well for text classification , spam detection , sentment analysis etc

### It considers evry features as independent from each other even if they actually might be dependent or related

#### Bayes Theorem:

#### p(A|B) = p(B|A)*p(A)/p(B)

##### P(A∣B) = Probability of A given B
##### P(A) = Prior probability of A
##### P(B∣A) = Likelihood
##### P(B) = Evidence

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.naive_bayes import CategoricalNB

data = {
    'Covid': ['Yes','No','Yes','No','Yes','No','Yes','Yes','No','No'],
    'Flu':   ['No','Yes','Yes','No','No','No','No','No','Yes','Yes'],
    'Fever': ['Yes','Yes','Yes','No','Yes','Yes','Yes','No','Yes','No']
}

df = pd.DataFrame(data)

for col in df.columns:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col])

X = df[['Covid', 'Flu']]
y = df['Fever']

model = CategoricalNB()
model.fit(X, y)

prediction = model.predict([[1,1]])

print("Prediction:", prediction)

if prediction[0] == 1:
    print("Fever = Yes")
else:
    print("Fever = No")
```
