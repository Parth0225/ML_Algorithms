## Lasso Regression

### Lasso regression shrinks coefficients as regularization technique.

### But it may remove some features and make there values zero

#### Cost=RSS+λ(∣b1​∣+∣b2​∣+∣b3​∣+...)

#### Lambda controls penalty 
#### if lambda is small= 0.01 , then almost no regularization
#### if lambda is large= 100 , strong regularization

```python
import numpy as np
from sklearn.linear_model import Lasso
from sklearn.preprocessing import StandardScaler

# 1. Create synthetic student data
np.random.seed(42)
n_samples = 100

# Useful features
study_hours = np.random.uniform(1, 10, n_samples)
sleep_hours = np.random.uniform(5, 9, n_samples)
attendance = np.random.uniform(50, 100, n_samples)

# Irrelevant noise features (these have nothing to do with grades)
shoe_size = np.random.uniform(5, 12, n_samples)
favorite_number = np.random.randint(1, 10, n_samples)

# Target: Marks depend on study hours, attendance, and sleep, plus random noise
marks = 15 + (6.5 * study_hours) + (0.4 * attendance) + (1.2 * sleep_hours) + np.random.normal(0, 3, n_samples)

# Combine into a feature matrix X
X = np.column_stack([study_hours, sleep_hours, attendance, shoe_size, favorite_number])
feature_names = ['Study Hours', 'Sleep Hours', 'Attendance %', 'Shoe Size (Noise)', 'Favorite Number (Noise)']

# 2. Scale features 
# Crucial Step: Lasso penalizes the absolute size of coefficients, 
# so features with larger numerical ranges would be unfairly punished if unscaled.
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 3. Train Lasso Regression model
# alpha controls the strength of regularization. Higher alpha = more coefficients driven to zero.
alpha_value = 1.5
lasso = Lasso(alpha=alpha_value)
lasso.fit(X_scaled, marks)

# 4. View the resulting coefficients
print(f"--- Lasso Coefficients (Alpha = {alpha_value}) ---")
for name, coef in zip(feature_names, lasso.coef_):
    print(f"{name:25}: {coef:.4f}")

print(f"\nIntercept: {lasso.intercept_:.2f}")
```
